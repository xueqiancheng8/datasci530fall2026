Try it yourself!

Create a machine learning pipeline

- Run a support vector machine regression to predict user ratings
- What is the out-of-sample performance on the test data?
- Use your data to predict reviews that were missing
- How does the distribution of missing to non-missing reviews compare?



````python
# Write your own code

from sklearn.pipeline import Pipeline
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score

# 1. Pipeline: bag of words -> truncated SVD -> support vector regression
#    (fit only on the training text, so the test set stays unseen)
svr_pipeline = Pipeline([
    ("bag_of_words", CountVectorizer(min_df=1, stop_words=custom_stop_words)),
    ("svd",          TruncatedSVD(n_components=50, random_state=42)),
    ("svr",          SVR(kernel="rbf"))
])
svr_pipeline.fit(text_train, y_train)

# 2. Out-of-sample performance on the test data
y_pred = svr_pipeline.predict(text_test)
rmse     = np.sqrt(mean_squared_error(y_test, y_pred))
r2       = r2_score(y_test, y_pred)
# Baseline: always guess the average training rating
rmse_avg = np.sqrt(mean_squared_error(y_test, np.full(len(y_test), y_train.mean())))
print(f"Test RMSE: {rmse:.3f}  (baseline guessing the mean: {rmse_avg:.3f})")
print(f"Test R^2:  {r2:.3f}")

# 3. Predict ratings for reviews that were missing a rating
data_missing = data[data["reviews.rating"].isnull()]
predicted_missing = svr_pipeline.predict(data_missing["reviews.text"])

# 4. Compare the distributions of observed vs. predicted ratings
comparison = pd.DataFrame({
    "Observed (non-missing)": data_nonmissing["reviews.rating"].describe(),
    "Predicted (missing)":    pd.Series(predicted_missing).describe()
})
print(comparison)

bins = np.arange(0.75, 5.5, 0.5)
plt.hist(data_nonmissing["reviews.rating"], bins=bins, density=True,
         alpha=0.5, label="Observed (non-missing)")
plt.hist(predicted_missing, bins=bins, density=True,
         alpha=0.5, label="Predicted (missing)")
plt.xlabel("Rating")
plt.ylabel("Density")
plt.legend()
plt.show()

# Takeaways:
# - The model beats guessing the mean, but only modestly (R^2 around 0.1).
# - Predicted ratings for the missing reviews cluster around 4-5, with far
#   less spread than the observed ratings. SVR pulls predictions toward the
#   middle, so it almost never predicts 1s or 2s.
````
